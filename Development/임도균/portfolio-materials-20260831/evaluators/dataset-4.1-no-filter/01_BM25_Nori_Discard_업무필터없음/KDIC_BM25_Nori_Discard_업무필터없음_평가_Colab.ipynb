# BM25 Nori Discard (업무 필터 없음)

Apache Lucene `KoreanAnalyzer`(Nori)의 `discard` 모드로 한국어 형태소를 분석하고 BM25로 427개 청크를 검색한 뒤, 검색평가대상 121개 질문의 Top-10 결과를 Gold 청크와 비교합니다.

- 외부 API와 API 키를 사용하지 않습니다.
- GPU도 필요하지 않습니다.
- 기본 BM25와 점수식·파라미터·Gold·업무 필터는 같고 Nori 복합어 모드만 `discard`로 고정합니다.
- `discard`: 복합어 원형은 버리고 분해된 형태소만 검색에 사용.
- 최초 실행 시 Lucene Nori 구성요소 약 14MB를 내려받습니다.

## 준비 파일

1. `KDIC_BM25_Nori_Discard_평가기.zip`
2. `KDIC_output.zip`
3. `test_dataset_4.1.xlsx`

```python
%pip -q install -U "pandas==2.2.2" "openpyxl>=3.1,<4" "JPype1>=1.5,<2"

import json
import shutil
import subprocess
import sys
import zipfile
from pathlib import Path

import pandas as pd
from IPython.display import display

WORK_ROOT=Path('/content/kdic_bm25_nori_discard_evaluation')
EVALUATOR_ROOT=WORK_ROOT/'evaluator'
RESULT_ROOT=WORK_ROOT/'results'
WORK_ROOT.mkdir(parents=True,exist_ok=True)
print('환경 준비 완료:',WORK_ROOT)
print('Java 실행 파일:',shutil.which('java'))
if shutil.which('java') is None:
    raise RuntimeError('Java를 찾지 못했습니다. Colab 기본 런타임에서 다시 실행해 주세요.')
print('BM25-Nori 평가는 API 키와 GPU를 사용하지 않습니다.')
```

## 1. 파일 업로드

아래 셀을 실행하고 준비한 파일 3개를 한 번에 선택합니다.

```python
from google.colab import files
uploaded=files.upload()
for filename,content in uploaded.items():
    target=WORK_ROOT/filename
    target.write_bytes(content)
    print(f'업로드: {target.name} ({target.stat().st_size:,} bytes)')
```

```python
dataset_candidates=list(WORK_ROOT.glob('*.xlsx'))
kdic_candidates=[p for p in WORK_ROOT.glob('*.zip') if p.name=='KDIC_output.zip' or p.name.startswith('KDIC_output (')]
evaluator_candidates=[p for p in WORK_ROOT.glob('*.zip') if 'BM25' in p.name and 'Nori' in p.name and 'Discard' in p.name and '평가기' in p.name and '평가결과' not in p.name]
if not dataset_candidates: raise RuntimeError('평가 XLSX를 찾을 수 없습니다.')
if not kdic_candidates: raise RuntimeError('KDIC_output.zip을 찾을 수 없습니다.')
if not evaluator_candidates: raise RuntimeError('KDIC_BM25_Nori_Discard_평가기.zip을 찾을 수 없습니다.')
def choose_latest(items): return max(items,key=lambda p:p.stat().st_mtime)
DATASET_PATH=choose_latest(dataset_candidates); KDIC_ZIP_PATH=choose_latest(kdic_candidates); EVALUATOR_ZIP_PATH=choose_latest(evaluator_candidates)
if EVALUATOR_ROOT.exists(): shutil.rmtree(EVALUATOR_ROOT)
EVALUATOR_ROOT.mkdir(parents=True)
with zipfile.ZipFile(EVALUATOR_ZIP_PATH) as archive: archive.extractall(EVALUATOR_ROOT)
scripts=list(EVALUATOR_ROOT.rglob('evaluate_bm25_nori.py'))
if len(scripts)!=1: raise RuntimeError(f'평가 스크립트를 하나로 결정할 수 없습니다: {scripts}')
EVALUATOR_SCRIPT=scripts[0]
print('평가데이터셋:',DATASET_PATH.name); print('검색 데이터:',KDIC_ZIP_PATH.name); print('평가기:',EVALUATOR_ZIP_PATH.name)
```

## 2. 비교 조건 설정

기본 BM25와 공정하게 비교하기 위해 `k1=1.5`, `b=0.75`, `Top-K=10`을 유지합니다.

```python
K1=1.5
B=0.75
TOP_K=10
LUCENE_VERSION='9.12.2'
DECOMPOUND_MODE='discard'
print({'검색 방식':'BM25 + Lucene Nori (discard)','토큰화':'KoreanAnalyzer(Nori)','Lucene':LUCENE_VERSION,'복합어 모드':DECOMPOUND_MODE,'k1':K1,'b':B,'Top-K':TOP_K,'업무 필터':'Gold 업무 사전 필터'})
```

## 3. Dry-run

Nori를 실행하기 전에 질문·청크·Gold 연결을 검사합니다.

```python
dry_dir=WORK_ROOT/'results_dry'
cmd=[sys.executable,str(EVALUATOR_SCRIPT),'--dataset',str(DATASET_PATH),'--sheet-name','Sheet1','--kdic-zip',str(KDIC_ZIP_PATH),'--output-dir',str(dry_dir),'--top-k',str(TOP_K),'--k1',str(K1),'--b',str(B),'--lucene-version',LUCENE_VERSION,'--decompound-mode',DECOMPOUND_MODE,'--no-domain-filter','--dry-run']
subprocess.run(cmd,cwd=EVALUATOR_ROOT,check=True)
```

## 4. BM25-Nori Discard 전체 평가

처음 실행할 때 Lucene JAR를 내려받습니다. 출력에 `Nori 토큰화 확인`과 형태소 목록이 표시되면 정상입니다.

```python
RESULT_ROOT.mkdir(parents=True,exist_ok=True)
cmd=[sys.executable,str(EVALUATOR_SCRIPT),'--dataset',str(DATASET_PATH),'--sheet-name','Sheet1','--kdic-zip',str(KDIC_ZIP_PATH),'--output-dir',str(RESULT_ROOT),'--top-k',str(TOP_K),'--k1',str(K1),'--b',str(B),'--lucene-version',LUCENE_VERSION,'--decompound-mode',DECOMPOUND_MODE,'--no-domain-filter']
subprocess.run(cmd,cwd=EVALUATOR_ROOT,check=True)
```

## 5. 평가 결과 확인

```python
summary=json.loads((RESULT_ROOT/'summary.json').read_text(encoding='utf-8'))
overall=pd.DataFrame([summary['overall']]); by_domain=pd.read_csv(RESULT_ROOT/'summary_by_domain.csv'); questions=pd.read_csv(RESULT_ROOT/'question_results.csv')
print('검색 방식:',summary['retriever']); print('전체 평가 결과'); display(overall); print('도메인별 평가 결과'); display(by_domain); print('Hit@3 실패 질문 예시')
display(questions.loc[questions['hit_at_3']==0,['evaluation_id','question','domain','gold_chunk_ids','retrieved_chunk_ids']].head(10))
```
